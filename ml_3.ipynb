```python
import numpy as np
def df_to_x_y(df,window_size=5):
    df_to_np=df.to_numpy()
    x=[]
    y=[]
    
    for i in range(len(df_to_np)-window_size):
        row=[[a] for a in df_to_np[i:i+5]]

        x.append(row)
        label=df_to_np[i+5]
        y.append(label)
    return np.array(x),np.array(y)
```
